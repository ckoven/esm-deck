```python
import xarray as xr
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import linregress
```

```python
esm_metrics = pd.read_csv('esm_metrics_flat10.csv')
esm_timings = pd.read_csv('peak_years.csv')
esm_peak_amounts = pd.read_csv('peak_amounts.csv')
sink_fractions = pd.read_csv('sink_fractions.csv')
```

```python
g = pd.plotting.scatter_matrix(esm_timings, figsize=(15,15), marker = 'o', hist_kwds = {'bins': 10}, s = 60, alpha = 0.8, range_padding=1)
```

```python
esm_allinfo = pd.concat((esm_metrics, esm_peak_amounts, esm_timings, sink_fractions), axis=1)
```

```python
g = pd.plotting.scatter_matrix(esm_allinfo, figsize=(25,25), marker = 'o', hist_kwds = {'bins': 10}, s = 60, alpha = 0.8, range_padding=1)
```

```python
plt.plot(esm_allinfo['Land Sink Value'],esm_allinfo['tpw'], 'o')

plt.xlabel("Peak cumulative land sink (Pg C)")
plt.ylabel("Peak Warming Delay (yr)")
```

```python
slope, intercept, r_value, p_value, std_err = linregress(esm_allinfo['flat10 land frac'], esm_allinfo['tpw'])
print('r2', r_value**2)

xx = np.array([esm_allinfo['flat10 land frac'].min(), esm_allinfo['flat10 land frac'].max()])

yy = slope * xx + intercept


plt.plot(esm_allinfo['flat10 land frac'],esm_allinfo['tpw'], 'o')

plt.plot(xx, yy, color='grey')

plt.xlabel("flat10 Land Sink Fraction at year 100")
plt.ylabel("flat10-cdr Peak Warming Delay (yr)")
```

```python
slope, intercept, r_value, p_value, std_err = linregress(esm_allinfo['flat10 land frac'], esm_allinfo['zec50'])
print('r2', r_value**2)

plt.plot(esm_allinfo['flat10 land frac'],esm_allinfo['zec50'], 'o')

plt.xlabel("flat10 Land Sink Fraction at year 100")
plt.ylabel("ZEC50")
```
